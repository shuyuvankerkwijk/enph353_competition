```python
from google.colab import drive
drive.mount('/content/drive')

import cv2
import os
from google.colab.patches import cv2_imshow
```

```python
# Path to your cropped letter images
folder_path = "/content/drive/MyDrive/letter_crops"

# Get all .png files
image_files = sorted([
    f for f in os.listdir(folder_path)
    if f.endswith(".png") and f[0].isdigit()
])

# Track counts per label to avoid overwrite
label_counts = {}

for filename in image_files:
    full_path = os.path.join(folder_path, filename)

    # Load and display the image
    img = cv2.imread(full_path, cv2.IMREAD_GRAYSCALE)
    cv2_imshow(img)

    # Prompt for label
    print(f"Label for image {filename} (press a key and hit Enter): ")
    label = input().strip().upper()

    if not label:
        print("Skipped.")
        continue

    # Count how many times this label has been used
    count = label_counts.get(label, 0) + 1
    label_counts[label] = count

    # Rename the file
    new_filename = f"{label}_{count:03d}.png"
    new_path = os.path.join(folder_path, new_filename)

    os.rename(full_path, new_path)
    print(f"Renamed to {new_filename}")
```
